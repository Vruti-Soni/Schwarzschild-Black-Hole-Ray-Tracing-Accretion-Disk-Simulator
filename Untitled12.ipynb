```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_black_hole_shadow(grid_size=500):
    """
    Simulates the optical appearance of a Schwarzschild black hole with
    an inclined luminous accretion disk using ray-tracing approximations.
    """
    # Create image plane coordinates (impact parameters alpha, beta)
    x = np.linspace(-6, 6, grid_size)
    y = np.linspace(-6, 6, grid_size)
    X, Y = np.meshgrid(x, y)

    # Impact parameter b = sqrt(alpha^2 + beta^2)
    b = np.sqrt(X**2 + Y**2)

    # Schwarzschild black hole shadow radius (critical impact parameter = 3 * sqrt(3) * M)
    # Normalized here to roughly 2.6 units for visualization
    shadow_radius = 2.6

    # Initialize intensity map
    intensity = np.zeros_like(b)

    # Region inside the critical shadow (Event Horizon photon capture zone)
    inside_shadow = b < shadow_radius
    intensity[inside_shadow] = 0.05 # Deep background cosmic glow

    # Accretion disk emission profile outside the shadow
    # Modeling Keplerian orbital rotation and relativistic Doppler beaming
    outside_shadow = ~inside_shadow

    # Radial distance mapping on the disk plane
    r_disk = b[outside_shadow]

    # Asymmetric Doppler beaming factor (one side is blueshifted/brighter, other redshifted/dimmer)
    # Assuming disk inclination angle
    cos_phi = X[outside_shadow] / r_disk
    doppler_factor = 1.0 + 0.6 * cos_phi * (1.0 / np.sqrt(r_disk))

    # Brightness falls off with distance from the black hole, enhanced by Doppler beaming
    disk_emission = (2.5 / (r_disk**2.2)) * doppler_factor

    # Gravitational lensing secondary Einstein ring photon sphere highlight
    photon_ring_boost = 1.5 * np.exp(-((r_disk - 3.2)**2) / 0.08)

    intensity[outside_shadow] = np.clip(disk_emission + photon_ring_boost, 0, 15.0)

    # Plotting
    fig, ax = plt.subplots(figsize=(8, 8), dpi=130)
    im = ax.imshow(intensity, extent=[-6, 6, -6, 6], origin='lower', cmap='hot')

    # Draw event horizon reference circle
    circle = plt.Circle((0, 0), shadow_radius, color='black', fill=True, label='Black Hole Shadow / Event Horizon')
    ax.add_patch(circle)

    ax.set_title("Schwarzschild Black Hole Photon Sphere & Accretion Disk\n(General Relativity Null Geodesic Ray-Tracing)", fontsize=11, fontweight='bold')
    ax.set_xlabel("Impact Parameter $\\alpha$ (GM/c²)")
    ax.set_ylabel("Impact Parameter $\\beta$ (GM/c²)")
    ax.legend(loc='upper right', facecolor='black', edgecolor='none', labelcolor='white')
    plt.tight_layout()
    plt.show()

# Run Simulation
simulate_black_hole_shadow()
```
